# Classification: YOLO
- Used YOLOv11 for classification
- Dataset has to be saved in a specific structure (https://docs.ultralytics.com/pt/datasets/classify/)
- YOLO doesn't require a specific image size, it's all handled internally
- Runs on gpu/cpu

```python
# Run 'create_30k_ds.py' to create a ds with YOLO structure.
# This dataset has 30k real images and 30k fake images
# Split is 80% train, 10% test and 10% valid
# Structure:
'''
30kds/
│── train/
│   ├── Real/   (80% of real images)
│   ├── Fake/   (80% of selected fake images)
│
│── val/
│   ├── Real/   (10% of real images)
│   ├── Fake/   (10% of selected fake images)
│
│── test/
│   ├── Real/   (10% of real images)
│   ├── Fake/   (10% of selected fake images)
'''
```

```python
import os
import time
import random
import cv2
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
from ultralytics import YOLO
import torch

# Paths
model_path = "yolo11s-cls"
data_path = "C:/Users/danie/Desktop/UNI-4ano/DL/30kds"
save_path ="models"

# Training Parameters
epochs = 50
img_size = 512
batch_size = 64
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")
```

```python
def train_yolo(model_path, data_path, save_path, epochs, img_size, batch_size, device):
    """Trains YOLOv11 for image classification."""
    model = YOLO(model_path)
    model.train(
        project=save_path,
        name="yolo_classifier_no_weights",
        data=data_path,
        epochs=epochs,
        imgsz=img_size,
        batch=batch_size,
        device=device,
        plots=True,
        pretrained=False    # Remove or set to True. It's False just to compare pre-trained vs raw.
    )

train_yolo(model_path, data_path, save_path, epochs, img_size, batch_size, device)
```

```python
def classify_folder(model_path, folder_path, interval=2):
    """Classifies all images in a folder using YOLO and displays them with real & predicted labels."""
    
    model = YOLO(model_path)
    image_files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.jpg', '.png', '.jpeg'))]
    random.shuffle(image_files)

    for image_file in image_files:
        image_path = os.path.join(folder_path, image_file)
        img = cv2.imread(image_path)
        if img is None:
            continue
        
        # Run YOLO prediction
        results = model.predict(source=image_path)
        predicted_label = results[0].names[results[0].probs.top1]
    
        real_label = "Unknown"
        for class_name in ["Real", "Fake"]:
            if class_name.lower() in image_file.lower():
                real_label = class_name
                break
        
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        clear_output(wait=True)
        plt.figure(figsize=(6, 6))
        plt.imshow(img)
        plt.title(f"True Label: {real_label} | Predicted: {predicted_label}", fontsize=14)
        plt.axis("off")
        display(plt.gcf())
        plt.close()
        time.sleep(interval)

classify_folder(os.path.join("models", "yolo_classifier", "weights", "best.pt"), "class_test")
```
